## 📁 Vérification de l'arborescence projet

```python
import os
from pathlib import Path

# Racine du projet (remonter d'un niveau depuis notebooks/)
project_root = Path.cwd().parent

dossiers_attendus = [
    ".github/workflows",
    "data/raw",
    "data/processed", 
    "data/external",
    "docs",
    "notebooks",
    "src/dashboard/pages",
    "src/dashboard/utils",
    "src/dashboard/assets",
    "src/data",
    "src/models",
    "src/utils",
    "tests",
]

print("=" * 50)
print("📁 VÉRIFICATION ARBORESCENCE")
print("=" * 50)

tous_ok = True
for dossier in dossiers_attendus:
    chemin = project_root / dossier
    statut = "✅" if chemin.exists() else "❌"
    if not chemin.exists():
        tous_ok = False
    print(f"{statut} {dossier}")

print("=" * 50)
print("🎉 Arborescence complète !" if tous_ok else "⚠️ Certains dossiers manquent !")
```
